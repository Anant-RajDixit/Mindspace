### T_T_S


```python
from sklearn.model_selection import train_test_split
X_health=health.iloc[:,:-1]
Y_health=health['Mental_Health_Score']
X_health.head()
```

```python
one_hot=['Gender'	,'Country'	,'Academic_Level'	,'Most_Used_Platform',	'Purpose_Of_Use']
ss=['Age','Avg_Daily_Usage_Hours'	,'Daily_Unlocks'	,'Study_Hours'	,'Physical_Activity_Hours'	,'Sleep_Hours_Per_Night'	]
label=['Stress_Level']

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,OneHotEncoder,OrdinalEncoder
one_hot_pipe=Pipeline(steps=[('encode',OneHotEncoder(handle_unknown='ignore'))])
label_pipe=Pipeline(steps=[('Ordinal',OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))])
ss_pipe=Pipeline(steps=[('ss',StandardScaler())])

Pr=ColumnTransformer(transformers=[('one_hot_encoding',one_hot_pipe,one_hot),
                                ('Ordinal_Encoding',label_pipe,label)
                               ,('StandardScaler',ss_pipe,ss)])
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X_health,Y_health,test_size=0.25,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr_pipe=Pipeline(steps=[('preprocessing',Pr),('LinearRegression',LinearRegression())])
```

### BASE MODEL -- Linear Regression


```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error


lr_pipe.fit(X_train, Y_train)
lr_preds       = lr_pipe.predict(X_test)
lr_preds_train = lr_pipe.predict(X_train)

lr_r2_testing  = r2_score(Y_test, lr_preds)
lr_r2_training = r2_score(Y_train, lr_preds_train)
lr_mae         = mean_absolute_error(Y_test, lr_preds)

print(f"Accuracy of Training {lr_r2_training}")
print(f"Accuracy of Testing {lr_r2_testing}")
print(f'MAE {lr_mae}')
```

### Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor

rf_pipeline=Pipeline(steps=[('preprocessing',Pr),('RandomForest',RandomForestRegressor())])

rf_pipeline.fit(X_train, Y_train)
rf_preds          = rf_pipeline.predict(X_test)
rf_preds_training = rf_pipeline.predict(X_train)

rf_r2_testing  = r2_score(Y_test, rf_preds)
rf_r2_training = r2_score(Y_train, rf_preds_training)
rf_mae         = mean_absolute_error(Y_test, rf_preds)

print(f"Accuracy of Training {rf_r2_training}")
print(f"Accuracy of Testing {rf_r2_testing}")
print(f"MAE : {rf_mae}")
```

### ALthough it is performing goood but lets try hypertuned model and xgboost also

## Hypertuned Random Forest

```python
from sklearn.model_selection import RandomizedSearchCV

param_grid = {
    'RandomForest__n_estimators'       : [100,200,300],
    'RandomForest__max_depth'         : [5,10,15],
    'RandomForest__min_samples_split' : [2,5,10],
    'RandomForest__min_samples_leaf'  : [1,2,4]
}

random_search = RandomizedSearchCV(
    rf_pipeline,
    param_distributions=param_grid,
    n_iter = 15,
    cv = 5,
    scoring='r2',
    random_state=42,
    n_jobs =-1 #processor free
)

random_search.fit(X_train, Y_train)
```

```python
random_search.best_params_
```

```python
rf_best_pipeline = random_search.best_estimator_
rf_best_preds    = rf_best_pipeline.predict(X_test)

print(f"R2 Score for random forest after hyperparameter tuning {r2_score(Y_test, rf_best_preds)}")
print(f"MAE Score for random forest after hyperparameter tuning {mean_absolute_error(Y_test, rf_best_preds)}")
     
```

```python
from xgboost import XGBRegressor
from sklearn.pipeline import Pipeline

xgb_pipeline = Pipeline(steps=[
    ('preprocessing', Pr),
    ('XGBoost', XGBRegressor(
        objective='reg:squarederror',
        random_state=42,
        n_jobs=1
    ))
])

param_grid_xgb = {
    'XGBoost__n_estimators': [100, 200, 300, 500],
    'XGBoost__max_depth': [3, 4, 5, 7],
    'XGBoost__learning_rate': [0.01, 0.03, 0.05, 0.1],
    'XGBoost__subsample': [0.7, 0.8, 0.9, 1.0],
    'XGBoost__colsample_bytree': [0.7, 0.8, 0.9, 1.0],
    'XGBoost__min_child_weight': [1, 3, 5],
    'XGBoost__reg_alpha': [0, 0.1, 1],
    'XGBoost__reg_lambda': [1, 5, 10]
}
random_search_xgb = RandomizedSearchCV(
    estimator=xgb_pipeline,
    param_distributions=param_grid_xgb,
    n_iter=20,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1,
    verbose=1
)

xgb_pipeline.fit(X_train,Y_train)
random_search_xgb.fit(X_train, Y_train)

print("Best parameters:", random_search_xgb.best_params_)
print("Best CV R²:", random_search_xgb.best_score_)
```

```python
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

# Calculate RMSE for Linear Regression
lr_rmse = np.sqrt(mean_squared_error(Y_test, lr_preds))

# Calculate RMSE for default Random Forest
rf_rmse = np.sqrt(mean_squared_error(Y_test, rf_preds))

# Calculate metrics for tuned Random Forest
rf_tuned_preds = random_search.best_estimator_.predict(X_test)
rf_tuned_training_preds = random_search.best_estimator_.predict(X_train)

rf_tuned_rmse = np.sqrt(mean_squared_error(Y_test, rf_tuned_preds))
rf_tuned_mae = mean_absolute_error(Y_test, rf_tuned_preds)
rf_tuned_r2 = r2_score(Y_test, rf_tuned_preds)
r2_tuned_training = r2_score(Y_train, rf_tuned_training_preds)

# Calculate metrics for default XGBoost
xgb_preds = xgb_pipeline.predict(X_test)
xgb_preds_train = xgb_pipeline.predict(X_train)

xgb_rmse = np.sqrt(mean_squared_error(Y_test, xgb_preds))
xgb_mae = mean_absolute_error(Y_test, xgb_preds)
xgb_r2 = r2_score(Y_test, xgb_preds)
xgb_training_r2 = r2_score(Y_train, xgb_preds_train)

# Calculate metrics for tuned XGBoost
xgb_tuned_model = random_search_xgb.best_estimator_

xgb_tuned_preds = xgb_tuned_model.predict(X_test)
xgb_tuned_training_preds = xgb_tuned_model.predict(X_train)

xgb_tuned_rmse = np.sqrt(mean_squared_error(Y_test, xgb_tuned_preds))
xgb_tuned_mae = mean_absolute_error(Y_test, xgb_tuned_preds)
xgb_tuned_r2 = r2_score(Y_test, xgb_tuned_preds)
xgb_tuned_training_r2 = r2_score(Y_train, xgb_tuned_training_preds)

# Consolidate results
results = pd.DataFrame({
    'Model': [
        'Linear Regression',
        'Random Forest (default)',
        'Random Forest (tuned)',
        'XGBoost (default)',
        'XGBoost (tuned)'
    ],
    'R2': [
        lr_r2_testing,
        rf_r2_testing,
        rf_tuned_r2,
        xgb_r2,
        xgb_tuned_r2
    ],
    'Training R2': [
        lr_r2_training,
        rf_r2_training,
        r2_tuned_training,
        xgb_training_r2,
        xgb_tuned_training_r2
    ],
    'MAE': [
        lr_mae,
        rf_mae,
        rf_tuned_mae,
        xgb_mae,
        xgb_tuned_mae
    ],
    'RMSE': [
        lr_rmse,
        rf_rmse,
        rf_tuned_rmse,
        xgb_rmse,
        xgb_tuned_rmse
    ]
})

print(results.round(4))
```

```python
import joblib


best_xgb = random_search_xgb.best_estimator_

# Save the complete pipeline
joblib.dump(best_xgb, 'xgb_tuned_model.joblib')

print("Tuned XGBoost model saved successfully!")
```
